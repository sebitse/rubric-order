# Pilot readiness and main-v2 handoff
This notebook performs read-only planning unless the explicit report switch is enabled.

```python
from pathlib import Path

from rubric_order.api import (
    estimate_experiment,
    find_project_root,
    load_experiment_config,
    readiness_report,
)

PROJECT_ROOT = find_project_root(Path.cwd())
```

```python
PILOT_CONFIG = PROJECT_ROOT / "configs" / "pilot_v2.yaml"
MAIN_CONFIG = PROJECT_ROOT / "configs" / "main_v2.yaml"
DATA_PATH = PROJECT_ROOT / "HelpSteer2"
PILOT_RUN_ID = "pilot-002"
MAIN_RUN_ID = "main-v2"
WRITE_READINESS_REPORT = False
EXECUTE_MODEL_CALLS = False
MAX_CALLS = 0
```

```python
pilot_config = load_experiment_config(PILOT_CONFIG, DATA_PATH)
main_config = load_experiment_config(MAIN_CONFIG, DATA_PATH)
print(
    {
        "pilot_run_id": PILOT_RUN_ID,
        "pilot_calls": estimate_experiment(pilot_config)["total_inference_calls"],
    }
)
print(
    {
        "main_run_id": MAIN_RUN_ID,
        "main_calls": estimate_experiment(main_config)["total_inference_calls"],
    }
)
```

```python
pilot_dir = PROJECT_ROOT / "runs" / PILOT_RUN_ID
if pilot_dir.exists():
    display(readiness_report(pilot_config, PILOT_RUN_ID, write_report=WRITE_READINESS_REPORT))
else:
    print(
        "pilot-002 does not exist. Main remains blocked until judge approval and pilot readiness."
    )
```
